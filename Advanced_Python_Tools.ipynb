# **Assignment 1 - Working with NumPy**

**1. Create a 1D NumPy array with integers from 1 to 20. Perform the following operations:**

  a. Calculate the sum, mean, median, and standard deviation of the elements in the array.  
  b. Find the indices of elements greater than 10 in the array.


````python
import numpy as np

# Create a 1D NumPy array with integers from 1 to 20
arr = np.arange(1, 21)

# a. Calculate sum, mean, median, and standard deviation
arr_sum = np.sum(arr)
arr_mean = np.mean(arr)
arr_median = np.median(arr)
arr_std = np.std(arr)

# b. Find indices of elements greater than 10
indices_greater_than_10 = np.where(arr > 10)

# Display results
print("Array:", arr)
print("Sum:", arr_sum)
print("Mean:", arr_mean)
print("Median:", arr_median)
print("Standard Deviation:", arr_std)
print("Indices of elements > 10:", indices_greater_than_10)
````

**2. Create a 2D NumPy array of shape 4 X 4 with numbers ranging from 1 to 16.**
  
  a. Print the array.  
  b. Find the transpose of the array.  
  c. Calculate the row-wise and column-wise sums of the array.  




````python
import numpy as np

# Create a 2D NumPy array of shape 4x4 with numbers from 1 to 16
arr = np.arange(1, 17).reshape(4, 4)

# a. Print the array
print("Original Array:\n", arr)

# b. Find the transpose of the array
arr_transpose = arr.T
print("\nTranspose of Array:\n", arr_transpose)

# c. Calculate row-wise and column-wise sums
row_sum = np.sum(arr, axis=1)     # Sum across columns (row-wise)
col_sum = np.sum(arr, axis=0)     # Sum across rows (column-wise)

print("\nRow-wise Sum:", row_sum)
print("Column-wise Sum:", col_sum)
````

**3. Create two 3 X 3 arrays filled with random integers between 1 and 20.**

  a. Perform element-wise addition, subtraction, and multiplication.  
  b. Compute the dot product of the two arrays.  




````python
import numpy as np

# Create two 3x3 arrays filled with random integers between 1 and 20
arr1 = np.random.randint(1, 21, size=(3, 3))
arr2 = np.random.randint(1, 21, size=(3, 3))

# Display arrays
print("Array 1:\n", arr1)
print("\nArray 2:\n", arr2)

# a. Element-wise addition, subtraction, and multiplication
add_result = arr1 + arr2
sub_result = arr1 - arr2
mul_result = arr1 * arr2

print("\nElement-wise Addition:\n", add_result)
print("\nElement-wise Subtraction:\n", sub_result)
print("\nElement-wise Multiplication:\n", mul_result)

# b. Compute the dot product of the two arrays
dot_result = np.dot(arr1, arr2)
print("\nDot Product (Matrix Multiplication):\n", dot_result)
````

**4. Reshape a 1D array of size 12 into a 3 X 4 2D array and slice the first two rows and last two columns.**

````python
import numpy as np

# Create a 1D array of size 12
arr = np.arange(1, 13)

# Reshape into a 3x4 2D array
arr_2d = arr.reshape(3, 4)

print("Original 2D Array:\n", arr_2d)

# Slice: first two rows and last two columns
sliced_arr = arr_2d[:2, -2:]

print("\nSliced Array (First two rows & last two columns):\n", sliced_arr)
````

# **Assignment 2 - Working with Pandas**

**1. Create a DataFrame with the following data:**
  

```
  data = {
      'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
      'Age': [24, 27, 22, 32, 29],
      'Department': ['HR', 'Finance', 'IT', 'Marketing', 'HR'],
      'Salary': [45000, 54000, 50000, 62000, 47000]
  }
```
  a. Print the first five rows of the DataFrame.  
  b. Get the summary statistics of the 'Age' and 'Salary' columns.  
  c. Calculate the average salary of employees in the 'HR' department.


````python
import pandas as pd

# Create the DataFrame
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
    'Age': [24, 27, 22, 32, 29],
    'Department': ['HR', 'Finance', 'IT', 'Marketing', 'HR'],
    'Salary': [45000, 54000, 50000, 62000, 47000]
}

df = pd.DataFrame(data)

# a. Print the first five rows of the DataFrame
print("First five rows of the DataFrame:\n\n", df.head())

# b. Get summary statistics of 'Age' and 'Salary'
print("\nSummary Statistics for 'Age' and 'Salary':\n\n", df[['Age', 'Salary']].describe())

# c. Calculate average salary of employees in the 'HR' department
avg_salary_hr = df[df['Department'] == 'HR']['Salary'].mean()
print("\nAverage Salary of HR Department:", avg_salary_hr)
````

**2. Add a new column, 'Bonus', which is 10% of the salary.**

````python
# Add a new column 'Bonus' = 10% of Salary
df['Bonus'] = df['Salary'] * 0.10

# Display updated DataFrame
print("DataFrame with Bonus column:\n", df)
````

**3. Filter the DataFrame to show employees aged between 25 and 30.**

````python
# Filter employees aged between 25 and 30 (inclusive)
filtered_df = df[(df['Age'] >= 25) & (df['Age'] <= 30)]

# Display the filtered DataFrame
print("Employees aged between 25 and 30:\n\n", filtered_df)
````

**4. Group the data by 'Department' and calculate the average salary for each department.**

````python
# Group by 'Department' and calculate average salary
avg_salary_by_dept = df.groupby('Department')['Salary'].mean()

# Display the result
print("Average Salary by Department:\n\n", avg_salary_by_dept)
````

**5. Sort the DataFrame by 'Salary' in ascending order and save the result to a new CSV file.**

````python
# Sort the DataFrame by 'Salary' in ascending order
sorted_df = df.sort_values(by='Salary', ascending=True)

# Display the sorted DataFrame
print("DataFrame sorted by Salary (ascending):\n\n", sorted_df)

# Save the result to a new CSV file
sorted_df.to_csv('sorted_employees.csv', index=False)
print("\n Sorted DataFrame has been saved as 'sorted_employees.csv'")
````

# **Assignment 3 - Working with Matplotlib**

**1. Create a simple line plot for the following data:**
  ```
  x = [1, 2, 3, 4, 5]
  y = [10, 15, 25, 30, 50]
  ```
  a. Plot the data.  
  b. Customize the plot by adding a title, axis labels, and a grid.


````python
import matplotlib.pyplot as plt

# Data
x = [1, 2, 3, 4, 5]
y = [10, 15, 25, 30, 50]

# a. Plot the data
plt.plot(x, y, marker='o', linestyle='-', color='b')

# b. Customize the plot
plt.title("Simple Line Plot")          # Add title
plt.xlabel("X-axis")                   # Label for x-axis
plt.ylabel("Y-axis")                   # Label for y-axis
plt.grid(True)                         # Add grid

# Display the plot
plt.show()
````

**2. Create a bar graph to represent the marks scored by students in a subject:**  
  ```
  students = ['John', 'Jane', 'Alice', 'Bob']
  marks = [75, 85, 60, 90]
  ```
  a. Plot the data as a bar graph.  
  b. Customize the colors and add a title.


````python
import matplotlib.pyplot as plt

# Data
students = ['John', 'Jane', 'Alice', 'Bob']
marks = [75, 85, 60, 90]

# a. Plot the data as a bar graph
plt.bar(students, marks, color=['skyblue', 'orange', 'lightgreen', 'salmon'])

# b. Customize the plot
plt.title("Marks Scored by Students in a Subject")  # Add title
plt.xlabel("Students")                              # X-axis label
plt.ylabel("Marks")                                 # Y-axis label
plt.grid(axis='y', linestyle='--', alpha=0.7)       # Add horizontal grid lines

# Display the plot
plt.show()
````

**3. Create a pie chart to represent the percentage distribution of a company’s revenue from different regions:**  
  ```
  regions = ['North America', 'Europe', 'Asia', 'Others']
  revenue = [45, 25, 20, 10]
  ```
  a. Create a pie chart with the region names as labels.  
  b. Highlight the region with the highest revenue.


````python
import matplotlib.pyplot as plt

# Data
regions = ['North America', 'Europe', 'Asia', 'Others']
revenue = [45, 25, 20, 10]

# a. Create a pie chart with labels
# b. Highlight (explode) the region with the highest revenue
explode = [0.1 if r == max(revenue) else 0 for r in revenue]  # highlight highest revenue

plt.pie(
    revenue,
    labels=regions,
    autopct='%1.1f%%',     # show percentage values
    startangle=140,        # rotate chart for better look
    explode=explode,       # highlight top region
    shadow=True            # add a shadow effect
)

# Add title
plt.title("Company Revenue Distribution by Region")

# Display the pie chart
plt.show()
````

**4. Generate a histogram to show the frequency distribution of randomly generated integers between 1 and 100 (sample size = 1000).**

````python
import numpy as np
import matplotlib.pyplot as plt

# Generate random integers between 1 and 100 (sample size = 1000)
data = np.random.randint(1, 101, size=1000)

# Create the histogram
plt.hist(data, bins=10, color='skyblue', edgecolor='black')

# Customize the plot
plt.title("Frequency Distribution of Random Integers (1–100)")
plt.xlabel("Value Range")
plt.ylabel("Frequency")
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Display the histogram
plt.show()
````

# **Dictionaries**


comma seprated key-value pairs enclosed within {}



```
{key1: value1, key2: value2, .........}
```



````python
groceries = {'milk': 60, 'biscuits': 20, 'rice': 90, 'bread': 30}
print(groceries)
print(type(groceries))
print(len(groceries))
````

````python
# indexing

# print(groceries[0])
# KeyError: 0

print(groceries['milk'])
print(groceries['biscuits'])
````

````python
# updating dict value
groceries = {'milk': 60, 'biscuits': 20, 'rice': 90, 'bread': 30}
print(groceries)
groceries['milk'] = 70
print(groceries)
````

````python
groceries = {'milk': 60, 'biscuits': 20, 'rice': 90, 'bread': 30}

# print(groceries['sugar']) # KeyError: 'sugar'


groceries['sugar'] = 40 # adds new key-value pair to the dict
groceries['milk'] = 70 # updates the value of the key
print(groceries)
````

````python
student1 = {'maths': 80.5, 'eng': 76.0, 'phy': 89.0}

#fetch the marks for phy
print(student1['phy'])
# print(student1['chem']) # KeyError: 'chem'

#get()
print(student1.get('phy'))
print(student1.get('chem'))
""" If we use print[key] we get error but if we use get function it give None """
# if you want with some data you can do that too.
print(student1.get('hindi', 40))
````

````python
emp1 ={'id':1001, 'name': 'john', 'salary': 2000}
print(emp1.get('phone', 979692622))
print(emp1) # no change
print(emp1.get('id', 979692622)) # 1000 as it persent in the dictionary
````

````python
emp1 ={'id':1001, 'name': 'john', 'salary': 2000}

# membership operator = in, not in
print('id' in emp1)
print('phone' not in emp1)
print('john' in emp1)
# membership operator only check for the key not the value
````

````python
sem1_marks = {'maths': 78.5, 'eng': 71.0, 'phy': 86.5}
sem2_marks = {'maths': 81.5, 'bio': 90.5}

print(sem1_marks)
print(sem2_marks)
sem1_marks.update(sem2_marks)
print(sem1_marks)
````

````python
groceries1 = {'milk': 60, 'biscuits': 20, 'rice': 90, 'bread': 30}
groceries2 = {'rice': 110, 'sugar': 40, 'tea': 20}

groceries1.update(groceries2)
print(groceries1)

# pop()
groceries1.pop('sugar')
print(groceries1)
````

````python
groceries1 = {'milk': 60, 'biscuits': 20, 'rice': 90, 'bread': 30, 'milk':65}
print(groceries)
print(groceries1['milk'])

# Keys cannot be duplicated in a dict
````

````python
# d1 = {[1, 3, 5]: 9, [1, 2, 1]: 4}
# keys cannot be lists

d1 = {'a': 9, 'b': 4}
print(d1)

# not allowed keys- list, sets, dict => mutable datatypes
# allowed keys- str, int, float, bool, tuple => immutable datatypes

# it means keys of a dictionary can only be immutable datatypes!!!
# values can be any datatype

d2 = {'a': 9, 'b': 4.0, 'c':[1, 2, 4]}
print(d2)
print(d2['c'][1])
````

````python
student1 = { 'id': 101, 'name':'John', 'marks': {'maths': 78.5, 'eng': 71.0, 'phy': 86.5}}
# keys()
print(student1.keys(), type(student1.keys()))
# values()
print(student1.values(), type(student1.values()))
# items()
print(student1.items(), type(student1.items()))
````

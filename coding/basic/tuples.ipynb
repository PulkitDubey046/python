# **Concatenation, repetition, membership**

Count, intex
min, max sum

```python
student_details1 = (1001, "John")
student_details2 = (78.8, 91.0, 83.5, 79.5)

# + : concatenation operator
student_details = student_details1 + student_details2
print(student_details)
```

```python
# * : for repetition
student_details1 = (1001, "John")
student_details = student_details1 * 3
print(student_details)
```

```python
# in , not in
student_details2 = (78.8, 91.0, 83.5, 79.5)
print(78.8 in student_details2)
print(78.8 not in student_details2)
```

```python
# count
t1 = (10, 4, 1, 9, 0, 3, 1)
# tuple.index(element)
print(t1.index(4)) # what is the index of 4 in tuple t1

# print(t1.index(40))  # if a index is not there
# ValueError: tuple.index(x): x not in tuple

print(t1.index(1)) # if more element are present in tuple then index function will return the index of first element
```

```python
tuple1=(1, 8, 2, 5, 18, 219)

print(f"Biggest number: {max(tuple1)}")
print(f"Smallest number: {min(tuple1)}")
print(f"Sum of all numbers: {sum(tuple1)}")
```

# **Mutability & Immutability**

In Python, objects can be classified as either **mutable** or **immutable**. This refers to whether their state can be changed after they are created.

*   **Immutable Objects**: Once created, their value cannot be changed. If you try to modify an immutable object, a new object is created with the new value. Examples include numbers (integers, floats), strings, and tuples.

*   **Mutable Objects**: Their value can be changed after they are created without creating a new object. Examples include lists, dictionaries, and sets.

```python
# Demonstrating Immutable Objects (e.g., integers, strings, tuples)

x = 10
print(f"Initial x: {x}, id: {id(x)}")
x = x + 5 # A new integer object is created
print(f"Modified x: {x}, id: {id(x)}\n")

s = "hello"
print(f"Initial s: {s}, id: {id(s)}")
s = s + " world"
print(f"Modified s: {s}, id: {id(s)}\n")

t = (1, 2, 3)
print(f"Initial t: {t}, id: {id(t)}")
# t[0] = 10 # This would raise a TypeError: 'tuple' object does not support item assignment
print(f"Tuples are immutable: {t}, id: {id(t)}\n")


# Demonstrating Mutable Objects (e.g., lists, dictionaries)

l = [1, 2, 3]
print(f"Initial l: {l}, id: {id(l)}")
l.append(4) # The list object is modified in-place
print(f"Modified l: {l}, id: {id(l)}\n")

d = {'a': 1, 'b': 2}
print(f"Initial d: {d}, id: {id(d)}")
d['c'] = 3 # The dictionary object is modified in-place
print(f"Modified d: {d}, id: {id(d)}")
```

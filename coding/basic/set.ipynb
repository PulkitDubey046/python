# **Sets**
Sets are non-sequential collection of items.

Syntax: comma seprated elements enclosed within {}

Indexsing are not allowed.

```python
set1 = {10, "Python", 2.5}
print(set1)
print(type(set1))
```

```python
# print(set1[1])
# TypeError: 'set' object is not subscriptable
```

```python
# length of set1
print(len(set1))
```

```python
# Duplicate value are not allowed
set2 = {10, 20, 30, 20}
print(set2)
```

```python
nums ={1, 3, 2, 0, -1}

# Membership operator - in, not in

print(0 in nums)
print(10 not in nums)
```

```python
nums1={1, 3, 2, 0, -1}
nums2={3, 5}

#concatenation ?
# print(nums1 + nums2)
# TypeError: unsupported operand type(s) for +: 'set' and 'set'
```

```python
weekdays = ("Monday", "Tuesday", "Wednesday", "Thursday", "Friday")
print(weekdays)
print(type(weekdays))

weekdays_set = set(weekdays)
print(weekdays_set) # different order
print(type(weekdays_set))
```

```python
# Are sets mutable or immutable?
set1= {2, 0, -1}

# add()
set1.add(10)
print(set1)


# remove()
set1.remove(2)
print(set1)
```

**Operations on sets - union, intersection, difference**

```python
student1 = {"English", "Maths", "Cs", "Chemistry", "Pysics"}
student2 = {"English", "Maths", "Cs", "Economics", "History"}

print(student1, type(student1))
print(student2, type(student2))

# All subjects - union
print(student1.union(student2))
print(student1 | student2)

# Subjects taken by both students - intersection
print(student1.intersection(student2))
print(student1 & student2)

# Subjects taken by student1 but not student2 - difference
print(student1.difference(student2))
print(student1 - student2)
```

```python
days = {"Monday", "Tuesday", "Wednesday", "Thursday", "Friday"}
print(days)
print(type(days))

weekends = {"Saturday", "Sunday"}
print(weekends)
print(type(weekends))

# difference of sets
weekdays= days.difference(weekends)
print(weekdays)
```

# **Is my set frozen!**

```python
s1 = {1, 2, 5, 0}
print(s1)
s1.add(-1)
print(s1)   # It is mutable
```

```python
# Frozen sets - Immutable sets
fs1 = frozenset({10, 20, 30})
print(fs1)
print(type(fs1))

# fs1.add(40)
# AttributeError: 'frozenset' object has no attribute 'add'
```

```python
fs1 = frozenset({10, 20, 30})
fs2 = frozenset({20, 28})

print(fs1.union(fs2))
print(fs1.intersection(fs2))
print(fs1.difference(fs2))
```

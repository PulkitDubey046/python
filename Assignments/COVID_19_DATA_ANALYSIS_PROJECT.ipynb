```python
# Import the libraries
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# import the dataset
dataset = pd.read_csv('covid19_Confirmed_dataset.csv')
dataset.head()
```

```python
dataset.shape
```

```python
# delete the useless columns
df = dataset.drop(["Lat","Long"], axis= 1, inplace= True)
```

```python
dataset.head()
```

```python
corona_dataset_aggregated = dataset.groupby("Country/Region").sum()
# Drop the 'Province/State' column if it exists, as it's not relevant for numeric plotting
# and can cause TypeError if it contains non-numeric aggregated values.
if 'Province/State' in corona_dataset_aggregated.columns:
    corona_dataset_aggregated = corona_dataset_aggregated.drop(columns=['Province/State'])
```

```python
corona_dataset_aggregated.head()
```

```python
corona_dataset_aggregated.shape
```

```python
# visualise data related to a country
corona_dataset_aggregated.loc["China"].plot()
corona_dataset_aggregated.loc["India"].plot()
corona_dataset_aggregated.loc["Spain"].plot()
plt.legend()
```

```python
# calculate a good measure
corona_dataset_aggregated.loc["China"].plot()
```

```python
corona_dataset_aggregated.loc["China"][:3].plot()
```

```python
# calculate the first derivate of the curve
corona_dataset_aggregated.loc["China"].diff().plot()
```

```python
# maximum infection rate
corona_dataset_aggregated.loc["China"].diff().max()
```

```python
corona_dataset_aggregated.loc["India"].diff().max()
```

```python
corona_dataset_aggregated.loc["Spain"].diff().max()
```

```python
countries = list(corona_dataset_aggregated.index)
max_infection_rates = []

for c in countries:
  max_infection_rates.append(corona_dataset_aggregated.loc[c].diff().max())

corona_dataset_aggregated["Max_infection_rates"] = max_infection_rates
```

```python
corona_dataset_aggregated
```

```python
# create a new dataframe
corona_data = pd.DataFrame(corona_dataset_aggregated["Max_infection_rates"])
```

```python
corona_data
```

**1. Create a DataFrame with the following data:**
  

```
  data = {
      'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
      'Age': [24, 27, 22, 32, 29],
      'Department': ['HR', 'Finance', 'IT', 'Marketing', 'HR'],
      'Salary': [45000, 54000, 50000, 62000, 47000]
  }
```
  a. Print the first five rows of the DataFrame.  
  b. Get the summary statistics of the 'Age' and 'Salary' columns.  
  c. Calculate the average salary of employees in the 'HR' department.


````python
import pandas as pd

# Create the DataFrame
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
    'Age': [24, 27, 22, 32, 29],
    'Department': ['HR', 'Finance', 'IT', 'Marketing', 'HR'],
    'Salary': [45000, 54000, 50000, 62000, 47000]
}

df = pd.DataFrame(data)

# a. Print the first five rows of the DataFrame
print("First five rows of the DataFrame:\n\n", df.head())

# b. Get summary statistics of 'Age' and 'Salary'
print("\nSummary Statistics for 'Age' and 'Salary':\n\n", df[['Age', 'Salary']].describe())

# c. Calculate average salary of employees in the 'HR' department
avg_salary_hr = df[df['Department'] == 'HR']['Salary'].mean()
print("\nAverage Salary of HR Department:", avg_salary_hr)
````

**2. Add a new column, 'Bonus', which is 10% of the salary.**

````python
# Add a new column 'Bonus' = 10% of Salary
df['Bonus'] = df['Salary'] * 0.10

# Display updated DataFrame
print("DataFrame with Bonus column:\n", df)
````

**3. Filter the DataFrame to show employees aged between 25 and 30.**

````python
# Filter employees aged between 25 and 30 (inclusive)
filtered_df = df[(df['Age'] >= 25) & (df['Age'] <= 30)]

# Display the filtered DataFrame
print("Employees aged between 25 and 30:\n\n", filtered_df)
````

**4. Group the data by 'Department' and calculate the average salary for each department.**

````python
# Group by 'Department' and calculate average salary
avg_salary_by_dept = df.groupby('Department')['Salary'].mean()

# Display the result
print("Average Salary by Department:\n\n", avg_salary_by_dept)
````

**5. Sort the DataFrame by 'Salary' in ascending order and save the result to a new CSV file.**

````python
# Sort the DataFrame by 'Salary' in ascending order
sorted_df = df.sort_values(by='Salary', ascending=True)

# Display the sorted DataFrame
print("DataFrame sorted by Salary (ascending):\n\n", sorted_df)

# Save the result to a new CSV file
sorted_df.to_csv('sorted_employees.csv', index=False)
print("\n Sorted DataFrame has been saved as 'sorted_employees.csv'")
````
